```python
listA=[]
while True:
  item=input("請輸入A倉文具，輸入'/'結束:")
  if item=="/":
    break
  listA.append(item)
listB=[]
while True:
  item=input("請輸入B倉文具，輸入'/'結束:")
  if item=="/":
    break
  listB.append(item)
listC=listA+listB
listC.append("膠水")
listC="/".join(listC)
print(f"二者合併清單:{listC}")
```

```python
listA=[]
while True:
  item=input("請輸入您要點的餐點，輸入'/'結束：")
  if item=="/":
    break
  listA.append(item)
  if item=="沙拉":
    listA.remove("沙拉")
    print("今日無沙拉，已自動移除。")
listA.append("冰淇淋")
listA=",".join(listA)
print(f"完整訂單清單:{listA}")
```
